```python
# Importing
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
data = pd.read_csv('/content/drive/MyDrive/top_5_polluted_indian_cities_comparison.csv')
data.head()
```

```python
data = data.sort_values(by='timestamp')
data.head(10)
```

ANALYZING TOP 5 POLLUTED CITIES

```python
# (Assuming your columns are named 'City' and 'AQI')
top_5_cities = data.groupby('city')['AQI_Value'].mean().sort_values(ascending=False).head(5).index.tolist()

print(f"The Top 5 Most Polluted Cities are: {top_5_cities}")

# 3. Filter the dataframe to ONLY include these 5 cities
df_top5 = data[data['city'].isin(top_5_cities)].copy()
print(f"Filtered dataset shape: {df_top5.shape}")
```

```python
df_top5.head()
```

STATISTICAL COMPARISON

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Set the style for clean, professional charts
sns.set_theme(style="whitegrid")

# 1. AQI Distribution (Boxplot)
# This shows the median, the spread (IQR), and those extreme "spike" days (outliers)
plt.figure(figsize=(12, 6))
sns.boxplot(x='city', y='AQI_Value', data=df_top5,
            order=['Gurugram', 'Delhi', 'Patna', 'Lucknow', 'Kolkata'],
            palette='Spectral')

plt.title('AQI Distribution Comparison: Top 5 Cities', fontsize=16, fontweight='bold')
plt.ylabel('AQI Value', fontsize=12)
plt.xlabel('City', fontsize=12)
plt.axhline(100, color='blue', linestyle='--', label='Satisfactory Limit (100)')
plt.axhline(300, color='red', linestyle='--', label='Severe Threshold (300)')
plt.legend()
plt.tight_layout()
plt.show()

# 2. AQI Category Breakdown (Stacked Bar Chart)
# This is the "Headline" visual: What % of the year is actually breathable?
cat_counts = df_top5.groupby(['city', 'AQI_Category']).size().unstack(fill_value=0)
# Convert to percentages
cat_pct = cat_counts.div(cat_counts.sum(axis=1), axis=0) * 100

# Reorder columns to follow the logical order (Good -> Severe)
ordered_labels = ['Good', 'Satisfactory', 'Moderate', 'Poor', 'Very Poor', 'Severe']
existing_labels = [l for l in ordered_labels if l in cat_pct.columns]
cat_pct = cat_pct[existing_labels]

# Plotting
cat_pct.plot(kind='bar', stacked=True, figsize=(12, 7),
             color=['#00e400', '#ffff00', '#ff7e00', '#ff0000', '#8f3f97', '#7e0023'])

plt.title('Air Quality Composition: % Time Spent in Each Category', fontsize=16, fontweight='bold')
plt.xlabel('City', fontsize=12)
plt.ylabel('Percentage of Time (%)', fontsize=12)
plt.legend(title='AQI Category', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

# 3. Print the Hard Numbers for your report
print("--- STATISTICAL SUMMARY (AQI VALUE) ---")
summary = df_top5.groupby('city')['AQI_Value'].agg(['mean', 'median', 'max', 'std']).sort_values(by='mean', ascending=False)
summary
```

TEMPORAL / SEASONAL COMPARISON

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Ensure 'timestamp' is a datetime object and extract Month/Year
# Note: Replace 'Date' with your actual timestamp column name if different
df_top5['timestamp'] = pd.to_datetime(df_top5['timestamp'])
df_top5['Month'] = df_top5['timestamp'].dt.month
df_top5['Month_Name'] = df_top5['timestamp'].dt.month_name()

# 2. Monthly Trend Line (Average AQI per month)
plt.figure(figsize=(14, 7))
monthly_trend = df_top5.groupby(['Month', 'city'])['AQI_Value'].mean().unstack()

# Sort months chronologically for the plot
monthly_trend.plot(marker='o', figsize=(14, 7), linewidth=2)

plt.title('Seasonal AQI Trend', fontsize=16, fontweight='bold')
plt.xlabel('Month', fontsize=12)
plt.ylabel('Average AQI', fontsize=12)
plt.xticks(range(1, 13), ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'])
plt.grid(True, alpha=0.3)
plt.legend(title='City', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()

# 3. The "Pollution Calendar" Heatmap
plt.figure(figsize=(12, 8))
heatmap_data = df_top5.pivot_table(values='AQI_Value', index='city', columns='Month', aggfunc='mean')

sns.heatmap(heatmap_data, annot=True, fmt=".0f", cmap='YlOrRd', cbar_kws={'label': 'Avg AQI'})

plt.title('Monthly Pollution Heatmap (Risk Assessment)', fontsize=16, fontweight='bold')
plt.xlabel('Month', fontsize=12)
plt.ylabel('City', fontsize=12)
plt.xticks(ticks=[i + 0.5 for i in range(12)], labels=['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'])
plt.tight_layout()
plt.show()
```

FEATURE IMPORTANCE SHIFT

```python
import pandas as pd
import lightgbm as lgb
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Setup the features (Focusing on the main pollutants to see the shift clearly)
# Using the core pollutants + your engineered features
features = [col for col in df_top5.columns if any(p in col for p in ['PM2.5', 'PM10', 'NO2', 'SO2', 'CO', 'O3'])]
target = 'AQI_Value'

importance_data = []

print("Analyzing Feature Importance Shift across cities...")

# 2. Loop through each city and get its unique importance profile
for city in ['Gurugram', 'Delhi', 'Patna', 'Lucknow', 'Kolkata']:
    city_df = df_top5[df_top5['city'] == city]

    # Train a quick model for this specific city
    model = lgb.LGBMRegressor(n_estimators=100, random_state=42, verbose=-1)
    model.fit(city_df[features], city_df[target])

    # Store the results
    for feat, imp in zip(features, model.feature_importances_):
        importance_data.append({'City': city, 'Feature': feat, 'Importance': imp})

df_importance_shift = pd.DataFrame(importance_data)

# 3. Visualize the Shift
plt.figure(figsize=(15, 8))
# We'll pivot the data to make a heatmap: Cities vs. Top Features
# Taking only top 15 features across all cities for clarity
top_features = df_importance_shift.groupby('Feature')['Importance'].mean().sort_values(ascending=False).head(15).index

df_pivot = df_importance_shift[df_importance_shift['Feature'].isin(top_features)].pivot(index='Feature', columns='City', values='Importance')

sns.heatmap(df_pivot, annot=True, cmap='YlGnBu', fmt='.0f')
plt.title('Spatial Analysis: Which Pollutants Drive AQI in Which City?', fontsize=16, fontweight='bold')
plt.tight_layout()
plt.show()
```

MODEL TRANSFERABILITY MATRIX

```python
import pandas as pd
import numpy as np
from sklearn.metrics import mean_squared_error, mean_absolute_error
import lightgbm as lgb
import xgboost as xgb

cities = ['Gurugram', 'Delhi', 'Patna', 'Lucknow', 'Kolkata']

# Initialize empty matrices
rmse_matrix = pd.DataFrame(index=cities, columns=cities)
mae_matrix = pd.DataFrame(index=cities, columns=cities)

print(">>> Generating Cross-City Transferability Matrices...")

# 1. Outer Loop: Pick a city to TRAIN on
for train_city in cities:
    print(f"Training Ensemble on {train_city}...")

    # Get training data for this city (75% of its data)
    train_df = df_top5[df_top5['city'] == train_city].sort_values('timestamp')
    train_split = int(len(train_df) * 0.75)

    X_train = train_df.drop(columns=['AQI_Value', 'AQI_Category', 'city', 'timestamp', 'Month_Name', 'state'], errors='ignore').iloc[:train_split]
    y_train = train_df['AQI_Value'].iloc[:train_split]

    # Train the Ensemble
    lgb_m = lgb.LGBMRegressor(n_estimators=100, verbose=-1)
    xgb_m = xgb.XGBRegressor(n_estimators=100)

    lgb_m.fit(X_train, y_train)
    xgb_m.fit(X_train, y_train)

    # 2. Inner Loop: Pick a city to TEST on
    for test_city in cities:
        test_df = df_top5[df_top5['city'] == test_city].sort_values('timestamp')
        test_split = int(len(test_df) * 0.75) # Testing on the same time-period slice (the last 20%)

        X_test = test_df.drop(columns=['AQI_Value', 'AQI_Category', 'city', 'timestamp', 'Month_Name', 'state'], errors='ignore').iloc[test_split:]
        y_test = test_df['AQI_Value'].iloc[test_split:]

        # Predict using the TRAINED city's model on the TEST city's data
        preds = (0.5 * lgb_m.predict(X_test)) + (0.5 * xgb_m.predict(X_test))

        # Calculate scores
        rmse_matrix.loc[train_city, test_city] = np.sqrt(mean_squared_error(y_test, preds))
        mae_matrix.loc[train_city, test_city] = mean_absolute_error(y_test, preds)

# Convert values to float for clean display
rmse_matrix = rmse_matrix.astype(float).round(2)
mae_matrix = mae_matrix.astype(float).round(2)

print("\n=== CROSS-CITY RMSE MATRIX (Row=Train, Col=Test) ===")
rmse_matrix
```
